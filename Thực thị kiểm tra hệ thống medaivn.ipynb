# Báo cáo kiểm tra thử hệ thống MEDAIVN.COM
**Người thực hiện:** Phạm Việt Anh  
**Ngày thực hiện:** 28/07/2026  
**Phạm vi thực hiện:** Kiểm thử các chức năng và kiểm thử giao diện trên môi trường máy tính.

---

## 1. Môi trường kiểm thử
 **Thiết bị:** Laptop Windows 11.  
 **Trình duyệt:** Google Chrome.

---

## 2. Kịch bản kiểm thử
1. **Truy cập Website:** Nhập và tìm kiếm `medaivn.com` vào trình duyệt web.  
2. **Kiểm tra thử chức năng chi tiết của 3 mục trò chuyện, tra cứu và nhận diện:**  
     **Mục "Bắt đầu trò chuyện":** Thử nêu ra các triệu chứng bệnh như "mệt người", "đau họng" và "ho khan" sau đó xem tư vấn từ chatbot.  
     **Mục "tra cứu":** Thử tra cứu về "ngải cứu" xem kết quả tra ra có chính xác không.  
    **Mục "nhận diện":** Thử lấy bức ảnh trên mạng chụp về cây ngải cứu để tra cứu bằng hình ảnh xem web có trả ra đúng kết quả không.  
3. **Kiểm tra hiệu năng của trang:** Sử dụng công cụ hỗ trợ Google PageSpeed Insights và cảm nhận thực tế khi sử dụng.

---

## 3. Kết quả và tổng kết đánh giá
### 3.1. Đánh giá chung
Giao diện web chuyên nghiệp, tốc độ truy cập nhanh, thao tác dễ dàng, các chức năng cơ bản hoạt động ổn định.

### 3.2. Chi tiết kết quả
 **Truy cập Website:** Trang chủ hiển thị đầy đủ, không bị lỗi font hay mất ảnh.  
 **Kiểm tra thử chức năng chi tiết của 3 mục trò chuyện, tra cứu và nhận diện:** Các mục tương ứng hiện ra đúng nội dung, không lỗi 404, không bị giật lag hay delay. Khi thử tra cứu thì ra đúng kết quả với thực tế.  
 **Kiểm tra hiệu năng của trang:**  
     Kết quả tổng quan: **56/100**.  
    Ưu điểm: Chỉ số SEO và Hỗ trợ tiếp cận đạt **100/100**.  
    Nhược điểm cần khắc phục: Chỉ số LCP (**7.7s**) cao, ảnh hưởng đến trải nghiệm người dùng khi chờ đợi trang tải nội dung chính. Chỉ số Total Blocking Time (**360 mili giây**) cao.

**Kiến nghị:** Cần tối ưu hóa lại kích thước hình ảnh và các tệp script để giảm thời gian phản hồi.
