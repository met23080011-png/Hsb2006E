# TÌM HIỂU VỀ TESTER

**Tester Định nghĩa:** Tester (chuyên viên kiểm tra phần mềm) là người có nhiệm vụ kiểm tra và đánh giá chất lượng các sản phẩm công nghệ, phát hiện các lỗi (bug) hoặc bất cứ sai sót nào liên quan đến sản phẩm.

---

## Công việc của 1 tester:

1. **Phân tích yêu cầu:** Trước khi test cần đọc tài liệu và tìm hiểu phần mềm hoạt động như nào và xác định phạm vi yêu cầu (Nếu có yêu cầu test tập trung vào phần nào).
2. **Lập kế hoạch:** Viết bản kế hoạch về các bước hay các nội dung sẽ chạy thử.
3. **Thực hành kiểm tra:** Thực tế trên sản phẩm công nghệ.
4. **Quản lý các lỗi đã tìm được:** Sau khi chạy thử thực tế theo kịch bản đã lập ra thì sẽ viết 1 báo cáo "bug repor" về bao gồm tên lỗi là gì, các bước thao tác tạo ra lỗi, hình ảnh và video minh chứng.
